# Step 3: Feature Engineering
Derive some features from the existing features and include some extra features in  our dataset so that we can derive meaningful or interesting insights from the data we have.

```python
parts = df["date"].str.split("-", n = 3, expand = True)
df["year"]= parts[0].astype('int')
df["month"]= parts[1].astype('int')
df["day"]= parts[2].astype('int')
df.head()
```

Encoding the days of the week as Weekend or weekday, as the either must have at least an effect on the requirements to fulfill the demands.

```python
from datetime import datetime

def weekend_or_weekday(year, month, day):
    d = datetime(year, month, day)
    return 1 if d.weekday() > 4 else 0

df['weekend'] = df.apply(lambda x: weekend_or_weekday(x['year'], x['month'], x['day']), axis=1)
df.head()
```

It is appropriate to add a column that indicate whether there was any holiday on a particular day or not.

```python
from datetime import date
import holidays

kenya_holidays = holidays.country_holidays('KE', years=[2013, 2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023])
df['holidays'] = df['date'].apply(lambda x: 1 if kenya_holidays.get(x) else 0)
df.head()
```

We can add some cyclical features

```python
df['m1'] = np.sin(df['month'] * (2 * np.pi / 12))
df['m2'] = np.cos(df['month'] * (2 * np.pi / 12))
df.head()
```

We add a column who value indicates which day of the week it is.

```python
def which_day(year, month, day):
    return datetime(year, month, day).weekday()

df['weekday'] = df.apply(lambda x: which_day(x['year'], x['month'], x['day']), axis=1)
df.head()
```

It is important to remove columns that will not be useful for the model.

```python
df.drop('date', axis=1, inplace=True)
```

# Step 5: Model Training
Separate the features and target variable and split them into training and testing data. The best perfprming model on the validation data is selected.

```python
features = df.drop(['sales', 'year'], axis=1)
target = df['sales'].values


X_train, X_val, Y_train, Y_val = train_test_split(features, target,
                                                  test_size = 0.05,
                                                  random_state=22)
X_train.shape, X_val.shape
```

Normalize the data before feeding it into the machine learning models to achieve stable and fast training.

```python
# Normalizing the features for stable and fast training.
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_val = scaler.transform(X_val)
```

Train the machine learning models and select the best out of them using the validation dataset.

```python
models = [LinearRegression(), XGBRegressor(), Lasso(), Ridge()]

for i in range(4):
    models[i].fit(X_train, Y_train)

    print(f'{models[i]} : ')

    train_preds = models[i].predict(X_train)
    print('Training Error : ', mae(Y_train, train_preds))

    val_preds = models[i].predict(X_val)
    print('Validation Error : ', mae(Y_val, val_preds))
    print()
```
